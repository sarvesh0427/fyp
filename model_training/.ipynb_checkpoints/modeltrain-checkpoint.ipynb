```python
import pandas as pd
```

```python
call = pd.read_csv('completed_dataset.csv')
```

```python
df = pd.concat([call.iloc[:1], call.iloc[1:].sample(frac=1, random_state=42)]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X = df.drop('Disease', axis=1)
y = df['Disease']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
Y = le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size= 0.3, random_state= 42)
```

### 1. Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model1 = RandomForestClassifier(n_estimators=1000,class_weight='balanced', random_state=42)
model1.fit(X_train,y_train)
y_pred1 = model1.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_pred1, y_test)
```

### 2. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model2 = LogisticRegression()
model2.fit(X_train,y_train)
y_pred2 = model2.predict(X_test)
```

```python
accuracy_score(y_pred2, y_test)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Accuracy scores
accuracies = [0.9438, 0.9108]
models = ['Random Forest', 'Logistic Regression']

# Plot
sns.barplot(x=models, y=accuracies, palette='viridis')
plt.ylabel('Accuracy Score')
plt.ylim(0.90, 1.0)
plt.title('Model Accuracy Comparison')
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Confusion matrix for Random Forest
cm_rf = confusion_matrix(y_test, y_pred1)
disp_rf = ConfusionMatrixDisplay(confusion_matrix=cm_rf, display_labels=model1.classes_)
disp_rf.plot(cmap='Blues')
plt.title("Random Forest - Confusion Matrix")
plt.show()

# Confusion matrix for Logistic Regression
cm_lr = confusion_matrix(y_test, y_pred2)
disp_lr = ConfusionMatrixDisplay(confusion_matrix=cm_lr, display_labels=model2.classes_)
disp_lr.plot(cmap='Greens')
plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

# Predict probabilities
y_proba_rf = model1.predict_proba(X_test)[:, 1]
y_proba_lr = model2.predict_proba(X_test)[:, 1]

# Compute ROC curve and AUC
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_proba_rf)
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_proba_lr)
auc_rf = auc(fpr_rf, tpr_rf)
auc_lr = auc(fpr_lr, tpr_lr)

# Plot ROC
plt.figure()
plt.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {auc_rf:.2f})")
plt.plot(fpr_lr, tpr_lr, label=f"Logistic Regression (AUC = {auc_lr:.2f})")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison')
plt.legend(loc='lower right')
plt.grid()
plt.show()
```
