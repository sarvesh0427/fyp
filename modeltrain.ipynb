```python
import pandas as pd
```

```python
call = pd.read_csv('completed_dataset.csv')
```

```python
df = pd.concat([call.iloc[:1], call.iloc[1:].sample(frac=1, random_state=42)]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X = df.drop('Disease', axis=1)
y = df['Disease']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
Y = le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size= 0.3, random_state= 42)
```

### 1. Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model1 = RandomForestClassifier(n_estimators=1000,class_weight='balanced', random_state=42)
model1.fit(X_train,y_train)
y_pred1 = model1.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_pred1, y_test)
```

### 2. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model2 = LogisticRegression()
model2.fit(X_train,y_train)
y_pred2 = model2.predict(X_test)
```

```python
accuracy_score(y_pred2, y_test)
```
